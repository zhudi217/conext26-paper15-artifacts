```python
import matplotlib.pyplot as plt
import numpy as np

# Set global font size
plt.rcParams.update({
    'font.family': 'Times New Roman',
    'font.size': 24
})

np.random.seed(42)
```

# Figure 14(a)

```python
x_feat = np.arange(1, 31)
x_cls = np.arange(1, 11)

x_feat_iisy = np.arange(1, 17)
x_feat_shapeshifter = np.arange(1, 28)
x_cls_iisy = np.arange(2, 7)
x_cls_shapeshifter = np.arange(2, 4)

# Create 4 pairs of data, each with 2 groups
data = {
    'Feature': {
        'IIsy': [109, 209, 309, 409, 509, 609, 709, 809, 909, 1009, 1109, 1209, 1309, 1409, 1509, 1609],
        'SS': [110, 212, 314, 416, 518, 620, 722, 824, 926, 1028,
                         1130, 1232, 1334, 1436, 1538, 1640, 1742, 1844, 1946,
                         2048, 2150, 2252, 2354, 2456, 2568, 2670, 2772]        # 27 features max
    },
    'Class': {
        'IIsy': [1104, 1109, 1117, 1133, 1165],       # 11 feature (1102), 2-6 classes
        'SS': [1120, 1126]
    }
}

# Create figure with 4 subplots in a row
fig, axes = plt.subplots(1, 2, figsize=(10, 3))

# Plot each subplot
for i, (ax, (title, groups)) in enumerate(zip(axes, data.items())):
    if i == 0:
        ax.plot(x_feat_iisy, groups['IIsy'], marker='o', label='IIsy', color='steelblue')
        ax.plot(x_feat_shapeshifter, groups['SS'], marker='s', label='SS', color='coral')
    else:
        ax.plot(x_cls_iisy, groups['IIsy'], marker='o', label='IIsy', color='steelblue')
        ax.plot(x_cls_shapeshifter, groups['SS'], marker='s', label='SS', color='coral')
    # ax.set_title(title, fontsize=24)

    if i == 0:
        ax.set_xlabel('# Feature', fontsize=24)
    else:
        ax.set_xlabel('# Classes', fontsize=24)

    # Only leftmost subplot gets y label
    if i == 0:
        ax.set_ylabel('# SRAM Entry', fontsize=24)

    # Keep only left and bottom spines, hide top and right
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_visible(True)
    ax.spines['bottom'].set_visible(True)

    ax.tick_params(axis='both', labelsize=24)

    ax.grid(True, axis='y', alpha=0.3)
    ax.grid(False, axis='x')

plt.tight_layout()
plt.show()
```
